```python
import numpy as np
from IPython.display import display, Markdown

math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

meanMath = np.mean(math)
meanPhysics = np.mean(physics)
meanEnglish = np.mean(english)

minMath = np.min(math)
minPhysics = np.min(physics)
minEnglish = np.min(english)

maxMath = np.max(math)
maxPhysics = np.max(physics)
maxEnglish = np.max(english)

names = ["Math", "Physics", "English"]
means = [meanMath, meanPhysics, meanEnglish]
best = names[np.argmax(means)]

table = f"""
__________________________________________________________
| Предмет  | Середній      | Мінімум      | Максимум     |
|__________|_______________|______________|______________|
| Math:    | {meanMath}    | {minMath}    | {maxMath}    |
| Physics: | {meanPhysics} | {minPhysics} | {maxPhysics} |
| English: | {meanEnglish} | {minEnglish} | {maxEnglish} |
|__________|_______________|______________|______________|
| Лідер:   | {best}                                      |
|__________|_____________________________________________|
"""

display(Markdown(table))
```
