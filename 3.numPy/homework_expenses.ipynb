```python
import numpy as np
from IPython.display import display, Markdown

arr = [1200, 1500, 800, 600, 1000, 400, 900]

total = np.sum(arr)
mean = np.mean(arr)
max = np.max(arr)
min = np.min(arr)
days = np.sum(arr > 1000)

table = f"""
______________________________
| Показники       | Значення |
|_________________|__________|
| Total:          |  {total} |
| Mean:           |  {mean}  |
| Max:            |  {max}   |
| Min:            |  {min}   |
| Days:           |  {days}  |
|_________________|__________|

"""
display(Markdown(table))
```
