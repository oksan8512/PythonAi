```python
from IPython.display import display, Markdown
import numpy as np

arr = [2, 5, 4, 6, 7, 3, 1]
mean = np.mean(arr)
max = np.max(arr)
min = np.min(arr)
differ = np.max(arr) - np.min(arr)

table = f"""
______________________________
| Показники       | Значення |
|_________________|__________|
| Mean:           |  {mean}  |
| Max:            |  {max}   |
| Min:            |  {min}   |
| Difference:     | {differ} |
|_________________|__________|

"""
display(Markdown(table))
```
